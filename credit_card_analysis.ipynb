```python
#Data cleaning => Standardized the column header

df.rename(columns={
    ' Applicant_ID  ' : 'APPLICANT ID' , 
    'APPLICANT_GENDER??' : 'APPLICANT GENDER' ,
    ' Owned_Car  ' : 'OWNED CAR' ,
    'OWNED_REALTY??' : 'OWNED REALITY' ,
    ' Total_Children  ' : 'TOTAL CHILDREN' ,
    'TOTAL_INCOME??' : 'TOTAL INCOME' ,
    ' Income_Type  ' : 'INCOME TYPE' ,
    'EDUCATION_TYPE??' : 'EDUCATION TYPE' ,
    ' Family_Status  ' : 'FAMILY STATUS' ,
    'HOUSING_TYPE??' : 'HOUSING TYPE' ,
    ' Owned_Mobile_Phone  ' : 'OWNED MOBILE PHONE' ,
    'OWNED_WORK_PHONE??' : 'OWNED WORK PHONE' ,
    ' Owned_Phone  ' : 'OWNED PHONE' ,
    'OWNED_EMAIL??' : 'OWNED EMAIL' ,
    ' Job_Title  ' : 'JOB TITLE' ,
    'TOTAL_FAMILY_MEMBERS??' : 'TOTAL FAMILY MEMBERS' ,
    ' Applicant_Age  ' : 'APPLICANT AGE' ,
    'YEARS_OF_WORKING??' : 'YEARS OF WORKING' ,
    ' Total_Bad_Debt  ' : 'TOTAL BAD DEBT' ,
    'TOTAL_GOOD_DEBT??' : 'TOTAL GOOD DEBT' ,
    ' Status  ' : 'STATUS'
} , inplace=True)



df.head(2)

df.info()
      
```

```python
#Data cleaning => looking for unique values in each colunm

print(df['APPLICANT ID'].unique())

```

```python
#Cleaning inconsistence values on full name
df['APPLICANT ID'].to_list()
```

```python
#Cleaning inconsistence value in Salary
# .astype() to convert string value to integer/float value
df['APPLICANT ID'] = (
    df['APPLICANT ID']
    .astype(str)
    .replace('unknown', 'nan', regex=True)
    .replace(r"[?,NG,,,??,#, ]", "", regex=True)
    .replace('', 'nan', regex=True)
    .replace('nan', np.nan)
    .astype(str)
    
)

df['APPLICANT ID'].to_list()

print(df['APPLICANT ID'].unique())
```

```python
#handling misisng values in 'APPLICANT ID' colunm
#master_dataset["Payment Date"] = master_dataset["Payment Date"].fillna(master_dataset["Payment Date"].median())
#master_dataset["Payment Date"]


df['APPLICANT ID'] = df['APPLICANT ID'].fillna(df['APPLICANT ID'].median())

df['APPLICANT ID']

df['APPLICANT ID'].to_list()

#print(df['APPLICANT ID'].unique())
```
